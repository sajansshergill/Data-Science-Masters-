```python
from sklearn.neighbors import KNeighborsRegressor
import numpy as np
R = [[5, 2], [6, 5], [4, 2], [3, 3], [4, 7], [3, 6], [1, 8]]
T = [4, 3, 3, 3, 2, 2, 1]
q = [2, 4]
```

a) According to the nearest neighbor algorithm where Euclidean L₂ distance is used,
what is the expected GPA if f1 = 2 and f2 = 4?

```python
def Minkowski(x, y, p):
    return np.sum(abs(x - y)**p)**(1/p)
```

```python
def NNR(R, T, q, p):
    D = []
    for ri in R:
        D.append(Minkowski(np.array(ri), np.array(q), p))
    return T[np.argmin(D)]
print(NNR(R, T, q, 2))
```

or using sklearn

```python
NNR = KNeighborsRegressor(n_neighbors=1)
NNR.fit(R,T)
print(NNR.predict([q]))
```

b) According to the (k = 3)-NN algorithm where Euclidean L₂ distance is used, what is
the expected GPA if f1 = 2 and f2 = 4?

```python
k3NNR = KNeighborsRegressor(n_neighbors=3)
k3NNR.fit(R,T)
print(k3NNR.predict([q]))
```

```python
def k_NNR(R, T, q, k, p):
    D = []
    for ri in R:
        D.append(Minkowski(np.array(ri), np.array(q), p))
    E = [i[0] for i in sorted(enumerate(D), key=lambda x:x[1])]
    Nidx = E[:k]
    prediction = 0
    for ni in Nidx:
        prediction += T[ni]
    return prediction/k
print(k_NNR(R, T, q, 3, 2))
```

d) According to the (s = 1) distance weighted (k = 3)-NN algorithm where Euclidean
L₂ distance is used, what is the expected GPA if f1 = 2 and f2 = 4?

```python
def dwk_NNR(R, T, q, k, p, s):
    D = []
    for ri in R:
        D.append(Minkowski(np.array(ri), np.array(q), p))
    E = [i[0] for i in sorted(enumerate(D), key=lambda x:x[1])]
    W = [0] * k
    for i in range(k):
        W[i] = 1/D[E[i]]**s
    tot = sum(W)
    W = [x / tot for x in W]
    prediction = 0
    for i in range(k):
        prediction += T[E[i]]*W[i]
    return prediction
print(dwk_NNR(R, T, q, 3, 2, 1))
```

or using sklearn

```python
dwk3NNR = KNeighborsRegressor(n_neighbors=3, weights='distance')
dwk3NNR.fit(R,T)
print(dwk3NNR.predict([q]))
```
