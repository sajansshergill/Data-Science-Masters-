b) Find the variance value, σ2(X1∼100) where μ(X1∼99) = 10 and σ2(X1∼99) = 12.

```python
def mean_online(xn, n, pm):
    """
    Computes the online (incremental) mean.
    
    Parameters:
    xn (float): The nth value (X[n])
    n (int): The number of samples (n > 0)
    pm (float): The mean of X[1:n-1]
    
    Returns:
    float: The updated mean after including xn.
    """
    return ((n - 1) * pm + xn) / n

def var_online(xn, n, pm, pv):
    """
    Computes the online (incremental) variance.

    Parameters:
    xn (float): The nth value (X[n])
    n (int): The number of samples (n > 0)
    pm (float): The mean of X[1:n-1]
    pv (float): The variance of X[1:n-1]

    Returns:
    float: The updated variance after including xn.
    """
    m = mean_online(xn, n, pm)
    return (((n - 1) * (pv + pm**2) + xn**2) / n) - m**2

var_online(10, 100, 10, 12)
```

e) Find the simple moving average value, μ(X91∼100) where μ(X90∼99) = 9.5

```python
def sma_online(xn, xk, pm, k):
    """
    Computes the online (incremental) Simple Moving Average (SMA).

    Parameters:
    xn (float): The new element to be added (X[n])
    xk (float): The first element in the previous window (X[n-k])
    pm (float): The mean of X[n-k:n-1]
    k (int): The size of the window

    Returns:
    float: The updated moving average after including xn and removing xk.
    """
    return pm + (xn - xk) / k

# Example usage
# sma_online(new_value, first_value_in_window, previous_mean, window_size)

sma_online(10, 5, 9.5, 10)
```

```python
def sma_online(xn, xk, pm, k):
    """
    Computes the online (incremental) Simple Moving Average (SMA).

    Parameters:
    xn (float): The new element to be added (X[n])
    xk (float): The first element in the previous window (X[n-k])
    pm (float): The mean of X[n-k:n-1]
    k (int): The size of the window

    Returns:
    float: The updated moving average after including xn and removing xk.
    """
    return pm + (xn - xk) / k

def smv_online(xn, xk, pm, pv, k):
    """
    Computes the online (incremental) Simple Moving Variance (SMV).

    Parameters:
    xn (float): The new element to be added (X[n])
    xk (float): The first element in the previous window (X[n-k])
    pm (float): The mean of X[n-k:n-1]
    pv (float): The variance of X[n-k:n-1]
    k (int): The size of the window

    Returns:
    float: The updated moving variance after including xn and removing xk.
    """
    m = sma_online(xn, xk, pm, k)
    return pv + (xn**2 - xk**2) / k - m**2 + pm**2

# Example usage:
# smv_online(new_value, first_value_in_window, previous_mean, previous_variance, window_size)

smv_online(10,5, 9.5, 4, 10)
```

g) Find the linear reverse rank weighted moving average value, μw(p=1)(X97∼100).

```python
import numpy as np

def wkave(X, W):
    """
    Computes the weighted moving average.

    Parameters:
    X (list or numpy array): The data vector.
    W (list or numpy array): The weight vector (should have size <= X).

    Returns:
    float: The weighted moving average.
    """
    n = len(X)
    k = len(W)
    
    # Normalize the weights to sum to 1
    W = np.array(W) / sum(W)
    
    wm = 0
    for i in range(k):
        wm += X[n - k + i] * W[i]
    
    return wm

# Example usage:

wkave([1, 2, 3, 4, 5], [0.1, 0.2, 0.3, 0.4])
```

i) Find the linear reverse rank weighted moving variance value, σ2
w(p=1)(X97∼100)

```python
import numpy as np

def wkave(X, W):
    """
    Computes the weighted moving average.

    Parameters:
    X (list or numpy array): The data vector.
    W (list or numpy array): The weight vector (should have size <= X).

    Returns:
    float: The weighted moving average.
    """
    n = len(X)
    k = len(W)
    
    # Normalize the weights to sum to 1
    W = np.array(W) / sum(W)
    
    wm = sum(X[n - k + i] * W[i] for i in range(k))
    
    return wm

def wkvar(X, W):
    """
    Computes the weighted moving variance.

    Parameters:
    X (list or numpy array): The data vector.
    W (list or numpy array): The weight vector (should have size <= X).

    Returns:
    float: The weighted moving variance.
    """
    n = len(X)
    k = len(W)
    
    # Compute the weighted mean
    wm = wkave(X, W)
    
    # Normalize the weights to sum to 1
    W = np.array(W) / sum(W)
    
    # Compute the weighted variance
    wv = sum(((X[n - k + i] - wm) ** 2) * W[i] for i in range(k))
    
    return wv

# Example usage:
wkvar([1, 2, 3, 4, 5], [0.1, 0.2, 0.3, 0.4])
```

k) Find the exponentially weighted mean value, μe(X1∼100) where μe(X1∼99) = 10 with
α = 0.2.

```python
def ewmean(pm, xn, a):
    """
    Computes the exponentially weighted mean.

    Parameters:
    pm (float): The mean of X[1:n-1]
    xn (float): The new element to be added (X[n])
    a (float): The alpha value (0 < a < 1/2)

    Returns:
    float: The updated exponentially weighted mean.
    """
    return (1 - a) * pm + a * xn

# Example usage:
ewmean(10,10, 0.2)
```

l) Find the exponentially weighted moving average value, μe(X91∼100) where μe(X90∼99) =
10 with α = 0.2.

```python
def ewma(pm, xn, xk, k, a):
    """
    Computes the Exponentially Weighted Moving Average (EWMA).

    Parameters:
    pm (float): The mean of X[1:n-1]
    xn (float): The new element to be added (X[n])
    xk (float): The first element in the previous window (X[n-k])
    k (int): The size of the window
    a (float): The alpha value (0 < a < 1/2)

    Returns:
    float: The updated EWMA.
    """
    return (1 - a) * pm + a * xn - a * ((1 - a) ** (k - 1)) * xk

# Example usage:
ewma(10, 10, 5, 10, 0.2)
```
