## Introduction to Data Analysis with Pandas

## PART I

**1.** Create a DataFrame `sd` from `StudentGrades.csv` dataset

```python
import pandas as pd
import numpy as np
```

```python
sd = pd.read_csv('Student Grades.csv')
```

**2.** Display a summary of the basic information about this DataFrame and its data.

```python
sd.info()
```

**3.** Return the first 3 rows of the DataFrame `sd`.

```python
print(sd.head(3))
```

**4.** Select just the 'semester' and 'grade' columns from the DataFrame `sd`.

```python
print(sd[['Semester', 'Grades']])
```

**5.** Select the data in rows [3, 4, 8] and in columns ['semester', 'grade'].

```python
print(sd.loc[[3, 4, 8], ['Semester', 'Grades']])
```

**6.** Select the students who have passed the class. A passing grade is 60.00 or higher. 

```python
passed_students = sd[sd['Grades'] >= 60]
print(passed_students)
```

**7.** Select the students who failed the course in 14_Fall. A failing grade is 59.99 or lower

```python
failed_14_fall = sd[(sd['Semester'] == '14_Fall') & (sd['Grades'] < 60)]
print(failed_14_fall)
```

**8.** Select the students who got a B+. B+ grade range is 87.00 to 89.99. Hint: You can use 'loc' attribute and & operator

```python
b_plus_students = sd[(sd['Grades'] >= 87) & (sd['Grades'] <= 89.99)]
print(b_plus_students)
```

**9.**  Change Student 5's grade to 78.00

```python
sd.loc[sd['Student_ID'] == 5, 'Grades'] = 78.00
print(sd.loc[sd['Student_ID'] == 5])
```

**10.** Reduce every student grade by half a point. 

```python
sd['Grades'] = sd['Grades'] - 0.5
print(sd.head())
```

## PART II 

**11.** What is the average of all grades for this course ?

```python
average_grade = sd['Grades'].mean()
print(f"Average grade: {average_grade}")
```

**11.5** What is the average grade for each semester. Hint: You can use 'groupby' and 'sort_values' methods.

```python
average_per_semester = sd.groupby('Semester')['Grades'].mean().sort_values()
print(average_per_semester)
```

**12.** How many students are there in each semester? You can use 'groupby' and 'sort_values' methods.

```python
students_per_semester = sd.groupby('Semester')['Student_ID'].count().sort_values()
print(students_per_semester)
```

**13.** Show a report that shows the number of students, average grade and the grade standard deviation per semester

```python
report = sd.groupby('Semester')['Grades'].agg(['count', 'mean', 'std'])
print(report)
```

**14.** What are the lowest 10 grades? Sort students by grade (from lowest to highest). 

```python
lowest_10_grades = sd.sort_values(by='Grades', ascending=True).head(10)
print(lowest_10_grades)
```

**15.**  What are the highest five grades? Sort students by grade (from lowest to highest).

```python
highest_5_grades = sd.sort_values(by='Grades', ascending=True).tail(5)
print(highest_5_grades)
```

**16.**  What are the lowest 3 grades for each semester? Sort by semester first and then grade. Hint: You can use 'groupby' and 'sort_values' methods.

```python
lowest_3_per_semester = sd.sort_values(by=['Semester', 'Grades']).groupby('Semester').head(3)
print(lowest_3_per_semester)
```

**17.** What are the highest 3 values for each semester? Sort by semester first and then grade. Hint: You can create a new variable to assign your sd DataFrame and than use 'groupby' and 'sort_values' methods.

```python
highest_3_per_semester = sd.sort_values(by=['Semester', 'Grades'], ascending=[True, False]).groupby('Semester').head(3)
print(highest_3_per_semester)
```

**18.** Which semester has the highest grade average ?

```python
highest_avg_semester = sd.groupby('Semester')['Grades'].mean().idxmax()
print(f"Semester with highest average grade: {highest_avg_semester}")
```

**19.** Find the average of 10 best students in each semester.

```python
top_10_avg_per_semester = sd.sort_values(by=['Semester', 'Grades'], ascending=[True, False]).groupby('Semester').head(10).groupby('Semester')['Grades'].mean()
print(top_10_avg_per_semester)
```

**20.** What is the average grade for failing students each semester ?

```python
failing_avg_per_semester = sd[sd['Grades'] < 60].groupby('Semester')['Grades'].mean()
print(failing_avg_per_semester)
```

## PART III

**21.** Find above average students. Filter the data to show students whose grades are higher than the average.

```python
overall_avg = sd['Grades'].mean()
above_avg_students = sd[sd['Grades'] > overall_avg]
print(above_avg_students)
```

**22.** Find students that perform above average of their semetester. In other words, show students that are greater than semester average. You can depict each semester separetly.

```python
# 22. Find students who perform above the average of their semester
semester_avg = sd.groupby('Semester')['Grades'].transform('mean')
above_avg_per_semester = sd[sd['Grades'] > semester_avg]

# Display each semester separately
for semester, group in above_avg_per_semester.groupby('Semester'):
    print(f"\nAbove-average students in {semester}:")
    print(group)
    
```
