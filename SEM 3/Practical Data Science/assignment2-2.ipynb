```python
def standardize_zipcode(zip_code):
    """Standardizes ZIP codes to be 5 digits."""
    if pd.isna(zip_code):
        return None  # Return None for missing values
    zip_str = str(zip_code).split('-')[0] # Handle potential extensions and convert to string
    if len(zip_str) >= 5:
        return zip_str[:5] # Take the first 5 digits if 5 or more
    if not zip_str.isdigit():
        return None # Mark str values as none
    else:
        # Pad with '10' followed by zeros to make it 5 digits
        return (zip_str + '10' + '0' * (5 - len(zip_str) - 2))[:5]

# Apply the function to the 'ZIP Code' column
df_NYC['ZIP Code'] = df_NYC['ZIP Code'].apply(standardize_zipcode)
df_NYC = df_NYC.dropna(subset=['ZIP Code'])
# Display the updated ZIP Code column and info to verify
display(df_NYC['ZIP Code'].head())
df_NYC['ZIP Code'].info()
```

```python
df_ver = df_NYC.copy() # Start with df_NYC as the base

def get_expected_borough(zip_code):
    """Determines the expected borough based on NYC ZIP code ranges."""
    if pd.isna(zip_code):
        return None # Return None for missing values

    try:
        # Ensure zip_code is treated as a string for consistent slicing/checking
        zip_str = str(zip_code).split('-')[0] # Handle potential extensions
        if not zip_str.isdigit() or len(zip_str) < 5:
             return None # Return None for non-digit or too short zip codes

        zip_int = int(zip_str[:5]) # Take the first 5 digits and convert to int

    except (ValueError, IndexError):
        return None # Return None for invalid formats

    if 10001 <= zip_int <= 10282:
        return 'Manhattan'  # Manhattan
    elif 10451 <= zip_int <= 10475:
        return 'Bronx'     # The Bronx
    elif 11201 <= zip_int <= 11239:
        return 'Brooklyn'  # Brooklyn
    elif 11000 <= zip_int <= 11697: # Adjusted Queens range to include more NYC zips
        return 'Queens'    # Queens
    elif 10301 <= zip_int <= 10314:
        return 'Staten Island' # Staten Island
    else:
        return None # Return None for ZIP codes outside these ranges

# Apply the function to create a new column with expected borough
df_ver['Expected Borough'] = df_ver['ZIP Code'].apply(get_expected_borough)

# Fill null 'Borough' values with 'Expected Borough'
df_ver['Borough'] = df_ver['Borough'].fillna(df_ver['Expected Borough'])

# Drop rows where 'Expected Borough' is still null (meaning ZIP code didn't map to an NYC borough)
df_ver.dropna(subset=['Expected Borough'], inplace=True)

# Display the count of nulls in the Borough and Expected Borough columns after imputation and dropping
print("Null record data for Borough after processing:")
print(df_ver['Borough'].isnull().sum())

# Display the shape of the dataframe after dropping rows and the column
print("\nShape of the DataFrame after dropping rows with no expected borough and the 'Expected Borough' column:")
print(df_ver.shape)


# Optional: Display a sample of the df_ver dataframe after correction
print("\nFirst 5 rows of df_ver after processing:")
display(df_ver.head())
```

```python

# Cell 1 — Imports

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, FunctionTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    roc_auc_score, average_precision_score, balanced_accuracy_score,
    confusion_matrix, classification_report, brier_score_loss,
    roc_curve, precision_recall_curve
)
from sklearn.calibration import calibration_curve
from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier
```

```python

# Cell 2 — Data prep (uses df_ver; keeps only the 3 features + target)

# Make a working copy
dfc = df_ver.copy()

# Ensure dates are datetime and (re)compute License_Duration (safe even if already present)
dfc['Initial Issuance Date'] = pd.to_datetime(dfc['Initial Issuance Date'], errors='coerce')
dfc['Expiration Date']       = pd.to_datetime(dfc['Expiration Date'], errors='coerce')
dfc['License_Duration']      = (dfc['Expiration Date'] - dfc['Initial Issuance Date']).dt.days

# Keep only required columns and drop NAs
use_cols = ['Business Category', 'Borough', 'License_Duration', 'Lapsed']
dfc = dfc[use_cols].dropna().copy()
dfc['Lapsed'] = dfc['Lapsed'].astype(int)

# Features / target
num_features = ['License_Duration']
cat_features = ['Business Category', 'Borough']
X = dfc[num_features + cat_features]
y = dfc['Lapsed']

# Train / test split (stratified)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=42
)
```

```python

# Cell 3 — Preprocessing (scale numeric, one-hot categoricals) and transform

numeric_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

preprocess = ColumnTransformer([
    ('num', numeric_pipe, num_features),
    ('ohe', OneHotEncoder(handle_unknown='ignore', min_frequency=0.01), cat_features),
], remainder='drop')

# Fit on train only
Z_train = preprocess.fit_transform(X_train)
Z_test  = preprocess.transform(X_test)

# Dense arrays for tree models (GB prefers dense); RF is fine with either
Ztr_dense = Z_train.toarray() if hasattr(Z_train, 'toarray') else Z_train
Zte_dense = Z_test.toarray()  if hasattr(Z_test,  'toarray')  else Z_test
```

```python

# Cell 4 — Models: Gradient Boosting & Random Forest (train)

gb_clf = GradientBoostingClassifier(
    n_estimators=400,
    learning_rate=0.05,
    max_depth=3,
    min_samples_leaf=20,
    random_state=42
)

rf_clf = RandomForestClassifier(
    n_estimators=300,
    max_depth=None,
    min_samples_leaf=10,
    class_weight='balanced',
    n_jobs=-1,
    random_state=42
)

# Train
gb_clf.fit(Ztr_dense, y_train)
rf_clf.fit(Ztr_dense, y_train)
```

```python

# Cell 5 — Evaluation helpers + Test evaluation + Overfitting check

def _scores(est, Z):
    """Return class-1 probabilities if available, else decision scores."""
    if hasattr(est, 'predict_proba'):
        return est.predict_proba(Z)[:, 1]
    elif hasattr(est, 'decision_function'):
        return est.decision_function(Z)
    else:
        raise AttributeError("Estimator must implement predict_proba or decision_function.")

def evaluate_cls(name, est, Z_te, y_te):
    p = _scores(est, Z_te)
    yhat = (p >= 0.5).astype(int)
    print(f"\n{name} — Test")
    print("ROC-AUC:", roc_auc_score(y_te, p))
    print("PR-AUC :", average_precision_score(y_te, p))
    print("Balanced Accuracy:", balanced_accuracy_score(y_te, yhat))
    print("Confusion Matrix:\n", confusion_matrix(y_te, yhat))
    print(classification_report(y_te, yhat, digits=3))

def overfit_report(name, est, Z_tr, y_tr, Z_te, y_te):
    p_tr = _scores(est, Z_tr)
    p_te = _scores(est, Z_te)
    tr_roc = roc_auc_score(y_tr, p_tr); te_roc = roc_auc_score(y_te, p_te)
    tr_pr  = average_precision_score(y_tr, p_tr); te_pr = average_precision_score(y_te, p_te)
    print(f"\n{name} — Train vs Test")
    print(f"train_roc: {tr_roc:.6f}")
    print(f"test_roc : {te_roc:.6f}")
    print(f"train_pr : {tr_pr:.6f}")
    print(f"test_pr  : {te_pr:.6f}")
    print(f"test_brier: {brier_score_loss(y_te, p_te):.6f}")
    print(f"ΔROC (train - test): {tr_roc - te_roc:.6f}")
    print(f"ΔPR  (train - test): {tr_pr - te_pr:.6f}")

# Test metrics
evaluate_cls("Gradient Boosting", gb_clf, Zte_dense, y_test)
evaluate_cls("Random Forest",     rf_clf, Zte_dense, y_test)

# Overfitting check
overfit_report("Gradient Boosting", gb_clf, Ztr_dense, y_train, Zte_dense, y_test)
overfit_report("Random Forest",     rf_clf, Ztr_dense, y_train, Zte_dense, y_test)
```

```python

# Cell 6 — 5-fold Cross-Validation (wrap preprocess inside Pipeline to avoid leakage)

to_dense = FunctionTransformer(lambda X: X.toarray() if hasattr(X, 'toarray') else X)

gb_cv = Pipeline([
    ('pre', preprocess),
    ('dense', to_dense),
    ('model', GradientBoostingClassifier(
        n_estimators=400, learning_rate=0.05, max_depth=3, min_samples_leaf=20, random_state=42
    ))
])

rf_cv = Pipeline([
    ('pre', preprocess),
    ('dense', to_dense),
    ('model', RandomForestClassifier(
        n_estimators=300, max_depth=None, min_samples_leaf=10,
        class_weight='balanced', n_jobs=-1, random_state=42
    ))
])

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

print("\n5-fold CV ROC-AUC (GB):", cross_val_score(gb_cv, X_train, y_train, cv=cv, scoring="roc_auc").mean())
print("5-fold CV PR-AUC  (GB):", cross_val_score(gb_cv, X_train, y_train, cv=cv, scoring="average_precision").mean())
print("\n5-fold CV ROC-AUC (RF):", cross_val_score(rf_cv, X_train, y_train, cv=cv, scoring="roc_auc").mean())
print("5-fold CV PR-AUC  (RF):", cross_val_score(rf_cv, X_train, y_train, cv=cv, scoring="average_precision").mean())
```

```python

# Cell 7 — Plots: ROC, Precision–Recall, Calibration, Confusion Matrices

p_gb = _scores(gb_clf, Zte_dense)
p_rf = _scores(rf_clf, Zte_dense)

# ROC
fpr_gb, tpr_gb, _ = roc_curve(y_test, p_gb)
fpr_rf, tpr_rf, _ = roc_curve(y_test, p_rf)
plt.figure(figsize=(6,5))
plt.plot(fpr_gb, tpr_gb, label=f"GB AUC={roc_auc_score(y_test, p_gb):.3f}")
plt.plot(fpr_rf, tpr_rf, label=f"RF AUC={roc_auc_score(y_test, p_rf):.3f}")
plt.plot([0,1],[0,1],'--',alpha=0.6)
plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate")
plt.title("ROC — Test"); plt.legend(); plt.tight_layout(); plt.show()

# Precision–Recall
prec_gb, rec_gb, _ = precision_recall_curve(y_test, p_gb)
prec_rf, rec_rf, _ = precision_recall_curve(y_test, p_rf)
plt.figure(figsize=(6,5))
plt.plot(rec_gb, prec_gb, label=f"GB AP={average_precision_score(y_test, p_gb):.3f}")
plt.plot(rec_rf, prec_rf, label=f"RF AP={average_precision_score(y_test, p_rf):.3f}")
plt.xlabel("Recall"); plt.ylabel("Precision")
plt.title("Precision–Recall — Test"); plt.legend(); plt.tight_layout(); plt.show()

# Calibration
frac_gb, mean_gb = calibration_curve(y_test, p_gb, n_bins=10, strategy='uniform')
frac_rf, mean_rf = calibration_curve(y_test, p_rf, n_bins=10, strategy='uniform')
plt.figure(figsize=(6,5))
plt.plot(mean_gb, frac_gb, 'o-', label=f"GB (Brier {brier_score_loss(y_test, p_gb):.3f})")
plt.plot(mean_rf, frac_rf, 'o-', label=f"RF (Brier {brier_score_loss(y_test, p_rf):.3f})")
plt.plot([0,1],[0,1],'--',alpha=0.6)
plt.xlabel("Mean predicted probability"); plt.ylabel("Observed fraction positives")
plt.title("Calibration — Test"); plt.legend(); plt.tight_layout(); plt.show()

# Confusion matrices at 0.50 threshold
from sklearn.metrics import ConfusionMatrixDisplay
fig, ax = plt.subplots(1,2, figsize=(10,4))
ConfusionMatrixDisplay.from_predictions(y_test, (p_gb>=0.5).astype(int), ax=ax[0])
ax[0].set_title("Gradient Boosting @ 0.50")
ConfusionMatrixDisplay.from_predictions(y_test, (p_rf>=0.5).astype(int), ax=ax[1])
ax[1].set_title("Random Forest @ 0.50")
plt.tight_layout(); plt.show()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.multioutput import MultiOutputRegressor
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error, r2_score

# --- 1) Copy and cleanup ---
df_problem2 = df_ver.copy()

# Drop irrelevant or non-geographic fields (per spec)
drop_cols = [
    "License Status", "Initial Issuance Date", "Expiration Date", "Business Name",
    "Lapsed", "License_Duration", "License Number", "Business Unique ID"
]
drop_cols = [c for c in drop_cols if c in df_problem2.columns]
df_problem2 = df_problem2.drop(columns=drop_cols, errors="ignore")

# --- 2) Predictors and targets ---
predictor_cols = [
    c for c in [
        "Borough", "NTA", "ZIP Code", "City", "State",
        "Business Category", "License Type"
    ] if c in df_problem2.columns
]
targets = ["Latitude", "Longitude"]

# ZIP as string
if "ZIP Code" in predictor_cols:
    df_problem2["ZIP Code"] = df_problem2["ZIP Code"].astype("string")

print("Predictor columns:", predictor_cols)
print("Target columns:", targets)
```

```python
# --- 3) Preprocessing ---
numeric_cols = [c for c in predictor_cols if pd.api.types.is_numeric_dtype(df_problem2[c])]
categorical_cols = [c for c in predictor_cols if c not in numeric_cols]

numeric_pipe = Pipeline([
    ("impute", SimpleImputer(strategy="median")),
    ("scale", StandardScaler())
])

categorical_pipe = Pipeline([
    ("impute", SimpleImputer(strategy="most_frequent")),
    ("ohe", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])

preprocessor = ColumnTransformer([
    ("num", numeric_pipe, numeric_cols),
    ("cat", categorical_pipe, categorical_cols)
])
```

```python
# --- 4) Split train vs predict ---
known_mask = df_problem2["Latitude"].notna() & df_problem2["Longitude"].notna()

X_known = df_problem2.loc[known_mask, predictor_cols]
y_known = df_problem2.loc[known_mask, targets]
X_missing = df_problem2.loc[~known_mask, predictor_cols]

print("Training rows:", X_known.shape[0])
print("Rows to impute:", X_missing.shape[0])
```

```python
# --- 5) Model and evaluation ---
base_model = RandomForestRegressor(
    n_estimators=400, random_state=42, n_jobs=-1
)
model = Pipeline([
    ("prep", preprocessor),
    ("reg", MultiOutputRegressor(base_model))
])

# 5-fold CV on known subset
kf = KFold(n_splits=5, shuffle=True, random_state=42)
maes, rmses, r2s = [], [], []

Xk = X_known.reset_index(drop=True)
yk = y_known.reset_index(drop=True)

for tr, te in kf.split(Xk):
    model.fit(Xk.iloc[tr], yk.iloc[tr])
    preds = model.predict(Xk.iloc[te])
    mae = mean_absolute_error(yk.iloc[te], preds)
    rmse = np.sqrt(np.mean((yk.iloc[te] - preds)**2))
    r2 = r2_score(yk.iloc[te], preds)
    maes.append(mae); rmses.append(rmse); r2s.append(r2)

metrics = {
    "MAE_mean": np.mean(maes), "RMSE_mean": np.mean(rmses), "R2_mean": np.mean(r2s)
}
print("Cross-validated metrics (joint model):", metrics)
```

```python
# --- 6) Fit final model and impute missing coordinates ---
model.fit(X_known, y_known)
preds = model.predict(X_missing)

df_problem2_imputed = df_problem2.copy()
df_problem2_imputed.loc[~known_mask, ["Latitude", "Longitude"]] = preds

print("Imputation complete.")
display(df_problem2_imputed.loc[~known_mask, predictor_cols + ["Latitude", "Longitude"]].head(10))
```

```python
# ===== Problem 2 Model Comparison (Multi-Output Regression) =====
import numpy as np
import pandas as pd
from sklearn.pipeline import Pipeline
from sklearn.multioutput import MultiOutputRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor, ExtraTreesRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error, r2_score

try:
    from xgboost import XGBRegressor
    has_xgb = True
except ImportError:
    has_xgb = False
    print("⚠️ XGBoost not installed — skipping XGBRegressor.")

# --- Prepare training data ---
Xk = X_known.reset_index(drop=True)
yk = y_known.reset_index(drop=True)

# --- Define candidate models ---
models = {
    "LinearRegression": LinearRegression(),
    "KNeighborsRegressor": KNeighborsRegressor(n_neighbors=10),
    "RandomForestRegressor": RandomForestRegressor(n_estimators=400, random_state=42, n_jobs=-1),
    "ExtraTreesRegressor": ExtraTreesRegressor(n_estimators=400, random_state=42, n_jobs=-1),
    "GradientBoostingRegressor": GradientBoostingRegressor(random_state=42),
}

if has_xgb:
    models["XGBRegressor"] = XGBRegressor(
        n_estimators=400, learning_rate=0.1, max_depth=6, subsample=0.8, colsample_bytree=0.8, n_jobs=-1, random_state=42
    )

# --- Evaluation function ---
def evaluate_model(name, model, X, y, preprocessor, n_splits=5):
    kf = KFold(n_splits=n_splits, shuffle=True, random_state=42)
    maes, rmses, r2s = [], [], []
    pipe = Pipeline([("prep", preprocessor), ("reg", MultiOutputRegressor(model))])

    for tr, te in kf.split(X):
        pipe.fit(X.iloc[tr], y.iloc[tr])
        preds = pipe.predict(X.iloc[te])
        maes.append(mean_absolute_error(y.iloc[te], preds))
        rmses.append(np.sqrt(np.mean((y.iloc[te] - preds) ** 2)))
        r2s.append(r2_score(y.iloc[te], preds))

    return {
        "MAE": np.mean(maes),
        "RMSE": np.mean(rmses),
        "R2": np.mean(r2s),
    }

# --- Run comparisons ---
results = {}
for name, model in models.items():
    print(f"Evaluating {name}...")
    metrics = evaluate_model(name, model, Xk, yk, preprocessor)
    results[name] = metrics

# --- Sort and display results ---
results_df = pd.DataFrame(results).T.sort_values("RMSE")
display(results_df.style.format("{:.4f}").set_caption("Model Comparison: Problem 2 (Geospatial Imputation)"))
```

```python
# Cell A — Setup, NYC bounds, and clean Problem 2 view

import numpy as np
import pandas as pd

# ---------- NYC sanity bounds ----------
NYC_LAT_MIN, NYC_LAT_MAX = 40.3, 41.1
NYC_LON_MIN, NYC_LON_MAX = -74.35, -73.40

def clip_coords(arr):
    """Clip [lat, lon] predictions to NYC bounds. Accepts (n,2) ndarray."""
    arr = np.asarray(arr, dtype=float).copy()
    arr[:, 0] = np.clip(arr[:, 0], NYC_LAT_MIN, NYC_LAT_MAX)
    arr[:, 1] = np.clip(arr[:, 1], NYC_LON_MIN, NYC_LON_MAX)
    return arr

def in_bounds_df(df):
    """Return boolean mask for rows whose (lat,lon) are within NYC bounds."""
    return (
        df["Latitude"].between(NYC_LAT_MIN, NYC_LAT_MAX)
        & df["Longitude"].between(NYC_LON_MIN, NYC_LON_MAX)
    )

# ---------- Build df_problem2 from df_ver ----------
df_problem2 = df_ver.copy()

# Drop irrelevant + Problem-1 artifacts + pure IDs (ignore if missing)
drop_cols = [
    "License Status","Initial Issuance Date","Expiration Date","Business Name",
    "Lapsed","License_Duration","License Number","Business Unique ID"
]
df_problem2 = df_problem2.drop(columns=[c for c in drop_cols if c in df_problem2.columns], errors="ignore")

targets = ["Latitude","Longitude"]
predictor_cols = [c for c in ["Borough","ZIP Code","NTA","City"] if c in df_problem2.columns]
for c in predictor_cols:
    df_problem2[c] = df_problem2[c].astype("string")

# Known / missing split, then hard-filter OOB training targets
known_mask_all = df_problem2["Latitude"].notna() & df_problem2["Longitude"].notna()
X_known_all = df_problem2.loc[known_mask_all, predictor_cols].reset_index(drop=True)
y_known_all = df_problem2.loc[known_mask_all, targets].reset_index(drop=True)

inb = in_bounds_df(y_known_all)
X_known = X_known_all.loc[inb].reset_index(drop=True)
y_known = y_known_all.loc[inb].reset_index(drop=True)

X_missing = df_problem2.loc[~known_mask_all, predictor_cols].reset_index(drop=True)

print(f"Predictors: {predictor_cols}")
print(f"Training rows (before/after bounds): {len(X_known_all)} / {len(X_known)}")
print(f"Rows to impute: {len(X_missing)}")
```

```python
# Cell B — FASTEST: Hierarchical centroid imputation (NTA → ZIP → Borough) with NYC clipping

# Choose hierarchy based on available columns
grp_levels = [
    [c for c in ["NTA"] if c in predictor_cols],
    [c for c in ["ZIP Code"] if c in predictor_cols],
    [c for c in ["Borough"] if c in predictor_cols],
]

y_global = y_known.mean()

# Build centroid tables on in-bounds known rows
centroids = []
for lvl in grp_levels:
    if not lvl:
        continue
    tab = pd.concat([X_known[lvl], y_known], axis=1).groupby(lvl, dropna=False)[targets].mean()
    centroids.append((lvl, tab))

def centroid_predict(X, tables, y_fallback):
    # X: DataFrame of predictors
    # tables: list of (lvl_cols, centroid_table) built from known rows
    # y_fallback: Series with index ["Latitude","Longitude"]
    n = len(X)
    out = np.full((n, 2), np.nan, dtype=float)
    remain = np.ones(n, dtype=bool)

    # try each hierarchy level in order
    for lvl, tab in tables:
        if not remain.any():
            break
        idx = np.where(remain)[0]
        # join on current level
        m = X.iloc[idx][lvl].merge(tab, on=lvl, how="left")[["Latitude","Longitude"]].to_numpy()
        got = ~np.isnan(m).any(axis=1)
        out[idx[got]] = m[got]
        remain[idx[got]] = False

    # per-column fallback to global mean
    lat_fb = float(y_fallback["Latitude"])
    lon_fb = float(y_fallback["Longitude"])
    nan_lat = np.isnan(out[:, 0])
    nan_lon = np.isnan(out[:, 1])
    if nan_lat.any():
        out[nan_lat, 0] = lat_fb
    if nan_lon.any():
        out[nan_lon, 1] = lon_fb

    # final clip to NYC bounds
    return clip_coords(out)
```

```python
# Cell C — FAST & BETTER: Centroid base + residuals with ExtraTrees (no OHE; frequency encoding), with clipping

# 1) Centroid tables using in-bounds known rows
grp_levels = [
    [c for c in ["NTA"] if c in predictor_cols],
    [c for c in ["ZIP Code"] if c in predictor_cols],
    [c for c in ["Borough"] if c in predictor_cols],
]

def build_centroid_tables(X, y, levels):
    tabs = []
    for lvl in levels:
        if not lvl:
            continue
        tabs.append((lvl, pd.concat([X[lvl], y], axis=1).groupby(lvl)[targets].mean()))
    return tabs

def centroid_lookup(X, tables, y_fallback):
    """
    Return centroid coords for each row in X using hierarchical tables.
    Per-column fallback to global mean. No clipping here; caller clips final preds.
    """
    n = len(X)
    out = np.full((n, 2), np.nan, dtype=float)
    remain = np.ones(n, dtype=bool)

    for lvl, tab in tables:
        if not remain.any():
            break
        idx = np.where(remain)[0]
        m = X.iloc[idx][lvl].merge(tab, on=lvl, how="left")[targets].to_numpy()
        got = ~np.isnan(m).any(axis=1)
        out[idx[got]] = m[got]
        remain[idx[got]] = False

    # per-column fallback to global means
    lat_fb = float(y_fallback["Latitude"])
    lon_fb = float(y_fallback["Longitude"])
    nan_lat = np.isnan(out[:, 0])
    nan_lon = np.isnan(out[:, 1])
    if nan_lat.any():
        out[nan_lat, 0] = lat_fb
    if nan_lon.any():
        out[nan_lon, 1] = lon_fb

    return out  # (n,2)

tables  = build_centroid_tables(X_known, y_known, grp_levels)
y_mean  = y_known.mean()
base_train = centroid_lookup(X_known, tables, y_mean)

# 2) Frequency encoder (tiny & fast) — no OHE explosion
from sklearn.base import BaseEstimator, TransformerMixin

class FreqEnc(BaseEstimator, TransformerMixin):
    def __init__(self, cols): self.cols = cols
    def fit(self, X, y=None):
        X = pd.DataFrame(X, columns=self.cols)
        self.maps_ = {c: X[c].value_counts(dropna=False).astype(float) for c in self.cols}
        return self
    def transform(self, X):
        X = pd.DataFrame(X, columns=self.cols).copy()
        for c in self.cols:
            X[c] = X[c].map(self.maps_[c]).fillna(0.0).astype(np.float32)
        return X.values

fe   = FreqEnc(predictor_cols)
X_fe = fe.fit_transform(X_known)
y_res = y_known.values - base_train  # residuals (Lat, Lon)

# 3) Train small ExtraTrees on residuals (fast holdout for speed)
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.multioutput import MultiOutputRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, r2_score

# FIXED: train_test_split returns 2 outputs per input → 6 outputs here
Xtr, Xte, ytr, yte, btr, bte = train_test_split(
    X_fe, y_res, base_train, test_size=0.2, random_state=42
)

reg = ExtraTreesRegressor(
    n_estimators=150,
    max_depth=20,
    min_samples_leaf=5,
    n_jobs=-1,
    random_state=42
)
model = MultiOutputRegressor(reg)
model.fit(Xtr, ytr)

# Evaluate on holdout (compose final coords = base + residuals, then clip)
pred_res = model.predict(Xte)
pred     = clip_coords(bte + pred_res)
true     = clip_coords(bte + yte)

mae  = float(np.mean(np.abs(true - pred)))
rmse = float(np.sqrt(np.mean((true - pred)**2)))
r2   = float(r2_score(true, pred))
print({"MAE": mae, "RMSE": rmse, "R2": r2})

# 4) Fit on all known & impute missing
model.fit(X_fe, y_res)

df_fast = df_problem2.copy()
if len(X_missing) > 0:
    base_missing = centroid_lookup(X_missing, tables, y_mean)
    Xmiss_fe     = fe.transform(X_missing)
    pred_missing = clip_coords(base_missing + model.predict(Xmiss_fe))
    df_fast.loc[~known_mask_all, targets] = pred_missing

df_problem2_imputed = df_fast
print("Residual + ExtraTrees imputation complete (with NYC clipping).")
```

```python
# Proper comparison with a hold-out test split
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np, pandas as pd

# Use the in-bounds known data for fair evaluation
X_train, X_test, y_train, y_test = train_test_split(
    X_known, y_known, test_size=0.2, random_state=42
)

# ---------- Model 1: Centroid baseline ----------
grp_levels = [
    [c for c in ["NTA"] if c in predictor_cols],
    [c for c in ["ZIP Code"] if c in predictor_cols],
    [c for c in ["Borough"] if c in predictor_cols],
]

def build_centroids(X, y, levels):
    tables = []
    for lvl in levels:
        if lvl:
            tables.append((lvl, pd.concat([X[lvl], y], axis=1)
                              .groupby(lvl)[["Latitude","Longitude"]].mean()))
    return tables

def centroid_predict(X, tables, y_fallback):
    out = np.full((len(X), 2), np.nan)
    remain = np.ones(len(X), bool)
    for lvl, tab in tables:
        if not remain.any(): break
        idx = np.where(remain)[0]
        m = X.iloc[idx][lvl].merge(tab, on=lvl, how="left")[["Latitude","Longitude"]].to_numpy()
        ok = ~np.isnan(m).any(axis=1)
        out[idx[ok]] = m[ok]; remain[idx[ok]] = False
    out[np.isnan(out)] = y_fallback.values[np.isnan(out)]
    return clip_coords(out)

centroids_train = build_centroids(X_train, y_train, grp_levels)
y_global = y_train.mean()
y_pred_centroid = centroid_predict(X_test, centroids_train, y_global)

# ---------- Model 2: Residual + ExtraTrees ----------
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.multioutput import MultiOutputRegressor

# frequency-encode for speed
fe = {c: X_train[c].value_counts(dropna=False).astype(float) for c in predictor_cols}
def freq_encode(df):
    df = df.copy()
    for c in predictor_cols:
        df[c] = df[c].map(fe[c]).fillna(0.0)
    return df.to_numpy(dtype=np.float32)

Xtr_fe, Xte_fe = freq_encode(X_train), freq_encode(X_test)
# centroid base for residuals
tables = build_centroids(X_train, y_train, grp_levels)
base_train = centroid_predict(X_train, tables, y_global)
base_test  = centroid_predict(X_test,  tables, y_global)

y_res_train = y_train.values - base_train

reg = MultiOutputRegressor(ExtraTreesRegressor(
    n_estimators=150, max_depth=20, min_samples_leaf=5, n_jobs=-1, random_state=42))
reg.fit(Xtr_fe, y_res_train)
y_pred_fast = clip_coords(base_test + reg.predict(Xte_fe))

# ---------- Compare ----------
def metrics(y_true, y_pred):
    return {
        "MAE":  mean_absolute_error(y_true, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_true, y_pred)),
        "R2":   r2_score(y_true, y_pred)
    }

m1 = metrics(y_test, y_pred_centroid)
m2 = metrics(y_test, y_pred_fast)

results = pd.DataFrame({"Centroid": m1, "Residual + ExtraTrees": m2}).T
display(results.style.format("{:.5f}").set_caption("Hold-Out Model Comparison"))

better = "Residual + ExtraTrees" if m2["RMSE"] < m1["RMSE"] else "Centroid"
print(f"\n✅ Better model based on RMSE: {better}")
```
